```python
import numpy as np
import pylab as plt
from matplotlib import rcParams
from scipy.integrate import odeint
```

```python
rcParams['figure.figsize'] = (20,10)
```

alfa - коэффициент рождаемости жертв

beta - вероятность смерти жертвы при встрече с хищником

gamma - естественная убыль популяции хищников

delta - коэффициент размножения сытого хищника

```python
def equilibrium(a,b,g,d):
    return (g/d,a/b)

def Lotka_Volterra(y,t,a,b,g,d):
    prey_pop,pred_pop = y
    new_prey_pop = a*prey_pop-b*prey_pop*pred_pop
    new_pred_pop = -g*pred_pop + pred_pop*prey_pop*d
    return np.array([new_prey_pop, new_pred_pop])
```

```python
init_pops = np.array([100,1])
a = 2/3
b = 4/3
g = 1
d = 1
t = np.linspace(0, 20, 1001)
sol = odeint(Lotka_Volterra, init_pops, t, args=(a, b,g,d))
plt.plot(t, sol[:, 0], 'b', label='prey')
plt.plot(t, sol[:, 1], 'g', label='predator')
plt.legend(loc='best')
plt.xlabel('t')
plt.grid()
plt.show()
```
