```python
import seaborn as sns
import json
import numpy as np
from matplotlib import rcParams
import pylab as plt
import os
```

```python
path_to_data = 'bin\Release\Pics'
```

```python
files = os.listdir(path_to_data)
files
```

```python
with open(os.path.join(path_to_data,'TempMap.json')) as inp:
    tmap = json.load(inp)
```

```python
to_draw = np.array(tmap['map']).reshape(tmap['Height'],tmap['Width'])
```

```python
rcParams['figure.figsize'] = (20,10)
sns.heatmap(to_draw,cmap='nipy_spectral')
plt.show()
```
