```python
import seaborn as sns
import json
import numpy as np
import imageio
from matplotlib import rcParams
import pylab as plt
import os
```

```python
path_to_data = 'bin\Release\Pics'
```

```python
files = os.listdir(path_to_data)
files
```

```python
with open(os.path.join(path_to_data,'map.json')) as inp:
    _map = json.load(inp)
```

```python
h = _map['Height']
w = _map['Width']
```

```python
to_draw_t = np.array(_map['TemperatureMap']['map']).reshape(h,w)
to_draw_m = np.array(_map['MoistureMap']['map']).reshape(h,w)
```

```python
rcParams['figure.figsize'] = (20,10)
print('temperature')
sns.heatmap(to_draw_t,cmap='nipy_spectral')
plt.show()
print('moisture')
sns.heatmap(to_draw_m,cmap='nipy_spectral')
plt.show()
print('biomes')
plt.imshow(imageio.imread(os.path.join(path_to_data,'Biome.png')))
```
